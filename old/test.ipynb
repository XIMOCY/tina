```python
import threading
import keyboard

stop_flag = False  # 用于控制输入的全局变量

# 监听快捷键
def listen_for_exit():
    global stop_flag
    keyboard.wait("ctrl+q")  # 监听 Ctrl+Q
    stop_flag = True  # 按下后修改标志位

def user_input_function():
    global stop_flag
    lines = ""

    # 启动监听线程
    listener = threading.Thread(target=listen_for_exit, daemon=True)
    listener.start()

    while not stop_flag:  # 监听标志位
        try:
            line = input()
            if stop_flag:  # 如果按了 Ctrl+Q，立即退出
                break
            lines += line.upper() + " "
        except EOFError:
            break  # 兼容 Ctrl+D（Linux/macOS）或 Ctrl+Z+Enter（Windows）

    return lines.strip()

if __name__ == "__main__":
    print("请输入内容，按 Ctrl+Q 退出：")
    result = user_input_function()
    print("\n输入结果：", result)
```
